```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom, poisson
```

```python
#Binomial Distribution
n = 10
p = 0.5

x = np.arange(0, n + 1)
binomial_prob = binom.pmf(x, n, p)
print("Total probability =", sum(binomial_prob))
print("Binomial probabilities:")
for i, prob in zip(x, binomial_prob):
    print(i, ":", prob)

plt.bar(x, binomial_prob)
plt.xlabel("Number of Successes")
plt.ylabel("Probability")
plt.title("Binomial Distribution")
plt.show()
```

```python
#Poisson Distribution
mu = 4

x = np.arange(0, 15)
poisson_prob = poisson.pmf(x, mu)

print("Poisson probabilities:")
for i, prob in zip(x, poisson_prob):
    print(i, ":", prob)

plt.bar(x, poisson_prob)
plt.xlabel("Number of Events")
plt.ylabel("Probability")
plt.title("Poisson Distribution")
plt.show()
```

```python
#compare both
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.bar(np.arange(11), binom.pmf(np.arange(11), 10, 0.5), color='green')
plt.title("Binomial Distribution")
plt.xlabel("Successes")
plt.ylabel("Probability")

plt.subplot(1, 2, 2)
plt.bar(np.arange(15), poisson.pmf(np.arange(15), 4),color='purple')
plt.title("Poisson Distribution")
plt.xlabel("Events")
plt.ylabel("Probability")

plt.tight_layout()
plt.show()
```
